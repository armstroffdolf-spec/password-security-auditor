```python

# 🛡️ AUDITOR EXPRESS DE SEGURIDAD DE CONTRASEÑAS

# 1. Le pedimos una contraseña al usuario en la pantalla
password = input("🔑 Escribe una clave para auditar su seguridad: ")

# 2. Inicializamos nuestro contador de puntos y la lista de fallos
puntos = 0
fallos = []

# 3. Evaluamos la regla 1: Longitud mínima de 8 caracteres
if len(password) >= 8:
    puntos += 1
else:
    fallos.append("Tiene menos de 8 caracteres (muy corta).")

# 4. Evaluamos la regla 2: Que contenga al menos una mayúscula
if any(caracter.isupper() for caracter in password):
    puntos += 1
else:
    fallos.append("Le faltan letras mayúsculas (ej. A, B, C).")

# 5. Evaluamos la regla 3: Que contenga al menos un número
if any(caracter.isdigit() for caracter in password):
    puntos += 1
else:
    fallos.append("No tiene números (ej. 1, 2, 3).")

# 6. Evaluamos la regla 4: Que contenga símbolos especiales
if any(not caracter.isalnum() for caracter in password):
    puntos += 1
else:
    fallos.append("No tiene símbolos especiales (ej. @, #, $, !).")

# 7. Diagnóstico e informe final
print("\n" + "=" * 35)
print("📊 RESULTADO DEL ANÁLISIS DE SEGURIDAD")
print("=" * 35)

if puntos == 4:
    print("🟢 ¡Excelente! Es una contraseña muy fuerte. (4/4)")
elif puntos == 3:
    print("🟡 Aceptable, pero puede mejorar. (3/4)")
else:
    print(f"🔴 ¡Cuidado! Es una contraseña muy débil. ({puntos}/4)")

# Si hubo fallos, los mostramos uno por uno
if fallos:
    print("\n💡 Recomendaciones para hacerla más segura:")
    for fallo in fallos:
        print(f" ❌ {fallo}")
```
